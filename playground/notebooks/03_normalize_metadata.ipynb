### Normalizing the metadata into seperate triples

Ziel ist es die Metadaten zu normalisieren und damit in einen Graph zu adden in seperaten Triples.

```python
import random

from rdflib.graph import Graph
import os


FILE_DIR_PATH = "govdata_random_rdf/"

random_rdf_file = random.choice(os.listdir(FILE_DIR_PATH))

print(f"Parsing RDF file: {random_rdf_file}")

g = Graph()

g.parse(os.path.join(FILE_DIR_PATH, random_rdf_file), format="xml")

metadata = []

for s, p, o in g.triples((None, None, None)):
    metadata.append((s, p, o))

for triple in metadata:
    print(triple)
```

Now we extract the accessURL from the triples and console log it

```python
from rdflib import Namespace


DCAT = Namespace("http://www.w3.org/ns/dcat#")

access_urls = []
csv_files = []

for s, p, o in g.triples((None, DCAT.downloadURL, None)):
    access_urls.append(o)
    print(o)
```
