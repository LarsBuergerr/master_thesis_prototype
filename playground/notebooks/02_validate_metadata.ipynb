### Validation of Metadata via SHACL Shapes


Ziel ist es die gefetchten Metadaten durch pyshacl zu validieren durch bereit gestellte SHACL-Shapes

```python
from pathlib import Path
from rdflib import Graph
from pyshacl import validate

SHAPES_FILE = Path("~/masterthesis/master_thesis_prototype/playground/scripts/data/shapes_DCAT-AP.de 2.0 - Spezifikation & Konventionen (2023-01-23).ttl").expanduser().resolve()
VOCAB_CACHE  = Path("~/masterthesis/master_thesis_prototype/playground/scripts/data/vocab_cache.ttl").expanduser().resolve()

shapes_graph = Graph()
shapes_graph.parse(SHAPES_FILE, format="ttl")
print(f"Shapes: {len(shapes_graph)} Triples")

DATA_DIR = Path("~/masterthesis/master_thesis_prototype/data/sample_2026-06-18_13-45").expanduser().resolve()

data_files = sorted(DATA_DIR.glob("non_geo_rest.rdf"))
if not data_files:
    raise FileNotFoundError(f"Keine .rdf Dateien in {DATA_DIR}")

for data_file in data_files:
    data_graph = Graph()
    data_graph.parse(data_file, format="xml")

    conforms, results_graph, results_text = validate(
        data_graph=data_graph,
        shacl_graph=shapes_graph,
        inference=None,
        abort_on_first=False,
        allow_infos=True,
        allow_warnings=True,
        meta_shacl=False,
        advanced=False,
        js=False,
        debug=False,
    )

    print("=" * 80)
    print(f"File: {data_file.name}  |  Conforms: {conforms}")
    print("=" * 80)
    if not conforms:
        print(results_text)
    print()
```
